```python
import random
import string
import pandas as pd
import random
from datetime import datetime, timedelta


def validate_customer_id(customer_id):
    pattern = r"^00k[a-zA-Z0-9]{15}$"
    
    if re.match(pattern, customer_id):
        return "PASS"
    else:
        return "FAIL"



def validate_oli_id(val):
    pattern = r"^OLI-\d{6}$"
    
    if re.match(pattern, val):
        return "PASS"
    else:
        return "FAIL"


from datetime import datetime

def validate_created_date(val):
    pattern = r"^(1[0-2]|[1-9])/(3[01]|[12][0-9]|[1-9])/\d{4}$"
    
    # Step 1: Check format
    if not re.match(pattern, val):
        return "FAIL"
    
    # Step 2: Check if it's a real calendar date
    try:
        month, day, year = map(int, val.split("/"))
        datetime(year, month, day)
        return "PASS"
    except:
        return "FAIL"


def validate_opp_id(val):
    pattern = r"^Opp-\d{6}$"
    
    if re.match(pattern, val):
        return "PASS"
    else:
        return "FAIL"


def validate_sub_delivery_org(val):
    allowed_values = ["analytics", "Data engineering", "", None]
    
    if val in allowed_values:
        return "PASS"
    else:
        return "FAIL"


def validate_stage(val):
    allowed_values = [
        "Discover", "Define", "On Bid", "Down Select",
        "Confirmed", "Signed Deal", "Lost", "Dropped",
        "Prediscover", "", None
    ]
    
    if val in allowed_values:
        return "PASS"
    else:
        return "FAIL"


def validate_domain(val):
    allowed_values = ["Hunting", "Mining"]
    
    if val in allowed_values:
        return "PASS"
    else:
        return "FAIL"


def validate_delivery_location(val):
    allowed_values = [
        "0", "Americas", "APAC", "Europe",
        "India", "Middle East", "South Africa",
        "", None
    ]
    
    if val in allowed_values:
        return "PASS"
    else:
        return "FAIL"


def validate_sales_region(val):
    allowed_values = [
        "Africa", "Asia", "Europe",
        "North America", "Oceania", "South America",
        "", None
    ]
    
    if val in allowed_values:
        return "PASS"
    else:
        return "FAIL"


def validate_tcv(val):
    pattern = r"^USD \d{6,7}\.\d{2}$"
    
    if re.match(pattern, val):
        return "PASS"
    else:
        return "FAIL"


def validate_currency(val):
    allowed_values = [
        "AUD", "CAD", "CNY", "EUR", "GBP",
        "INR", "JPY", "PHP",
        "US Dollar", "USD",
        "", None
    ]
    
    if val in allowed_values:
        return "PASS"
    else:
        return "FAIL"


def validate_forex_rate(val):
    pattern = r"^USD \d{1,3}\.\d{6}$"
    
    if re.match(pattern, val):
        return "PASS"
    else:
        return "FAIL"


def validate_contract_term(val):
    try:
        num = int(val)
        
        if 0 <= num <= 100:
            return "PASS"
        else:
            return "FAIL"
    except:
        return "FAIL"


def validate_deal_age(val):
    try:
        num = int(val)
        
        if 0 <= num <= 365:
            return "PASS"
        else:
            return "FAIL"
    except:
        return "FAIL"


import re
from datetime import datetime

def validate_datetime(val):
    pattern = r"^(1[0-2]|[1-9])/(3[01]|[12][0-9]|[1-9])/\d{4} ([01][0-9]|2[0-3]):[0-5][0-9]$"
    
    # Step 1: Format check
    if not re.match(pattern, val):
        return "FAIL"
    
    # Step 2: Real datetime check
    try:
        date_part, time_part = val.split(" ")
        month, day, year = map(int, date_part.split("/"))
        hour, minute = map(int, time_part.split(":"))
        
        datetime(year, month, day, hour, minute)
        return "PASS"
    except:
        return "FAIL"


def validate_country(val):
    allowed_values = [
        "USA", "India", "Germany", "France", "Australia",
        "Japan", "Brazil", "Canada", "UK", "South Africa",
        "", None
    ]
    
    if val in allowed_values:
        return "PASS"
    else:
        return "FAIL"


def validate_dmaic_check(val):
    allowed_values = ["True", "False", "", None]
    
    if val in allowed_values:
        return "PASS"
    else:
        return "FAIL"
```

```python
def validate_row(row):
    return {
        "oli_check": validate_oli_id(row["oli_id"]),
        "customer_check": validate_customer_id(row["customer_id"]),
        "opp_id_check": validate_opp_id(row["opp_id"]),
        "opp_created_date_check":validate_created_date(row["opp_created_date"]),
        "Sub_delivery_org_check":validate_sub_delivery_org(row["Sub_delivery_org"]),
        "stage_check":validate_stage(row["Stage"]),
        "domain_check":validate_domain(row["Hunting_Mining"]),
        "delivery_location_check":validate_delivery_location(row["delivery_location"]),
        "sales_region_check":validate_sales_region(row["sales_region"]),
        "tcv_check":validate_tcv(row["Tcv"]),
        "currency_check":validate_currency(row["Local currency"]),
        "forex_rate_check":validate_forex_rate(row["forex_rate"]),
        "contract_term_check":validate_contract_term(row["contract_term"]),
        "deal_age_check":validate_deal_age(row["deal_age"]),
        "datetime_check":validate_datetime(row["datetime"]),
        "country_check":validate_country(row["sales_country"]),
        "dmaic_check":validate_dmaic_check(row["dmaic_submission"])

        
    }
```
